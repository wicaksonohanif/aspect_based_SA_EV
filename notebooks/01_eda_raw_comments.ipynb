# 📊 Reading & Exploring Raw YouTube Comments Data

Notebook ini digunakan untuk membaca dan melakukan eksplorasi awal (*Exploratory Data Analysis* / EDA) terhadap berkas data mentah komentar YouTube hasil ekstraksi yang tersimpan di direktori `data/raw/`.

## 1. Import Libraries

```python
import glob
from pathlib import Path
import pandas as pd

# Set display options agar teks komentar tidak terpotong saat ditampilkan
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_columns", None)
```

## 2. Mencari Berkas Data Raw CSV Terakhir

```python
# Tentukan path folder data raw (relatif terhadap root atau notebook)
raw_data_dir = Path("../data/raw")
if not raw_data_dir.exists():
    raw_data_dir = Path("data/raw")

# Cari semua file CSV di folder raw
csv_files = sorted(raw_data_dir.glob("yt_comments_*.csv"))

if not csv_files:
    print("❌ Tidak ditemukan file CSV di folder data/raw/")
else:
    latest_csv = csv_files[-1]
    print(f"📁 Ditemukan {len(csv_files)} file CSV.")
    print(f"✅ Membaca file CSV terbaru: {latest_csv}")
```

## 3. Membaca CSV Menggunakan Pandas

```python
# Membaca data CSV dengan encoding utf-8-sig (sesuai spesifikasi SPEC-01)
df = pd.read_csv(latest_csv, encoding="utf-8-sig")

# Tampilkan dimensi dataset (jumlah baris dan kolom)
print(f"📊 Total Komentar: {len(df)} baris | Total Kolom: {df.shape[1]}")
df.head()
```

## 4. Inspeksi Ringkasan & Tipe Data

```python
# Informasi tipe data dan memory usage
df.info()
```

```python
# Cek apakah ada nilai kosong (missing values)
df.isnull().sum()
```
